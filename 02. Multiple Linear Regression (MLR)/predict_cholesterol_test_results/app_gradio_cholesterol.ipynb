```python
import gradio as gr
import joblib
import pandas as pd
```

```python
model = joblib.load('./cholesterol_model.pkl')
```

```python
def predict(bloody_type: str, smoker: bool, activity_level: str, age: int, weight: float, height: int):
    _smoker = 'Yes' if smoker else 'No'

    individual_prediction = {
        'bloody_type': bloody_type,
        'smoker': _smoker,
        'activity_level': activity_level,
        'age': age,
        'weight': weight,
        'height': height
    }

    predict_df = pd.DataFrame(individual_prediction, index=[1])
    cholesterol = model.predict(predict_df)

    return cholesterol[0]
```

```python
demo = gr.Interface(
    fn=predict,
    inputs=[
        gr.Radio(['A', 'B', 'AB', 'O']),
        'checkbox',
        gr.Radio(['Low', 'Moderate', 'High']),
        gr.Slider(20,80, step=1),
        gr.Slider(40,160, step=0.1),
        gr.Slider(150, 220, step=1)
    ],
    outputs=['number']
)
```

```python
demo.launch()
```
