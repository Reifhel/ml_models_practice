# Model Training

```python
X= df_countries.copy()

X.drop(columns=['country', 'income_category'], inplace=True)
```

```python
# Numeric features
numeric_features = ['child_mort', 'exports', 'health', 'imports', 'income', 'inflation', 'life_expec', 'total_fer', 'gdpp']

numeric_transformer = StandardScaler()
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer,numeric_features)
    ]
)
```

```python
X_transformed = preprocessor.fit_transform(X)
X_transformed
```

```python
# Covariance Matrix
np.cov(X_transformed)
```

```python
# Creating the PCA model
pca_model = PCA(n_components=3)
```

```python
X_pca = pca_model.fit_transform(X_transformed)
```

```python
# Creating a df with principal components
df_pca = pd.DataFrame(X_pca, columns=['PC1', 'PC2', 'PC3'])
df_pca.head(10)
```

```python
# Include the principal components on main df
df_countries['PC1'] = df_pca['PC1']
df_countries['PC2'] = df_pca['PC2']
df_countries['PC3'] = df_pca['PC3']

df_countries.head(10)
```

# Result Analysis

```python
# Eigenvalues
eigenvectors = pca_model.explained_variance_

# Eigenvectors
eigenvalues = pca_model.components_

print(f"Eigenvectors: \n{eigenvectors}\n")

print(f"Eigenvalues: \n{eigenvalues}")
```

```python
# Show 3D Charts with principal components
fig = px.scatter_3d(
                        df_countries, 
                        x='PC1', 
                        y='PC2', 
                        z='PC3', 
                        color='income_category', 
                        title='PCA Visualization', 
                        width=800, height=600, 
                        hover_data=['country']
                    )

fig.show()
```

```python
X_recovered = pca_model.inverse_transform(X_pca)
```

```python
X_recovered
```

```python
# Calculate the reconstruction error
reconstruction_error = mean_squared_error(X_transformed, X_recovered)
reconstruction_error
```
