# Exploratory Data Analysis

## Load the dataset

```python
df_score = pd.read_csv('./datasets/test_score.csv')
df_score.info()
```

## Understanding the dataset

```python
# Showing the stats from the dataset
df_score.describe()
```

```python
# Scatter plot to see the correlation
# X - hours_studied 
# Y - test_score
sns.scatterplot(data=df_score, x='hours_studied', y='test_score')
```

```python
# Checking for outliers
# Create a figure with 1 row and 2 columns
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# First plot on the left (index 0)
sns.boxplot(df_score, y='hours_studied', ax=axes[0])
axes[0].set_title("Outliers: Hours studied")

# Second plot on the right (index 1)
sns.boxplot(df_score, y='test_score', ax=axes[1])
axes[1].set_title("Outliers: Test Score")

plt.tight_layout() # Adjusts spacing to prevent overlap
plt.show()
```

```python
# Verifying the correlation with heatmap

# Create a figure with 1 row and 2 columns
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# First plot on the left (index 0)
sns.heatmap(data=df_score.corr('pearson'), annot=True, ax=axes[0])
axes[0].set_title("Correlation: Pearson")

# Second plot on the right (index 1)
sns.heatmap(data=df_score.corr('spearman'), annot=True, ax=axes[1])
axes[1].set_title("Correlation: Spearman")

plt.tight_layout() # Adjusts spacing to prevent overlap
plt.show()
```

```python
# Analysing the data distribution

# Create a figure with 1 row and 2 columns
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# First plot on the left (index 0)
sns.histplot(df_score, x='hours_studied', ax=axes[0])
axes[0].set_title("Distribution: Hours Studied")

# Second plot on the right (index 1)
sns.histplot(df_score, x='test_score', ax=axes[1])
axes[1].set_title("Distribution: Test Score")

plt.tight_layout() # Adjusts spacing to prevent overlap
plt.show()
```
